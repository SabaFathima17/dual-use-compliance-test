```python
import pandas as pd
df = pd.read_csv("data/compliance_data.csv", encoding ="utf-8-sig")
print(df.shape)
df.head()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
print(df.columns)
```

```python
df = df.rename(columns = {"unnamed:_0":"prompt_id"})
df.head()
```

```python
df["prompt_id"] = df["prompt_id"].str.strip().str.upper()
print(df["prompt_id"].head())
```

```python
model_columns = ["claude", "gemini", "deepseek", "perplexity","grok","chatgpt"]
print(df[model_columns].isna().sum())
print(sorted(df[model_columns].stack().unique()))
```

```python
df["category"] = df["category"].str.strip().str.replace(" ", "_").str.lower()
df["category"].head()
df["category"] = df["category"].str.replace("-", "_")
df["category"].unique()
df["category"] = df["category"].str.replace({"phising" : "phishing", "vhishing_script" : "vishing_script"})
df["category"].unique()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
os.makedirs("outputs", exist_ok = True)
```

```python
data = df[model_columns].copy()
data.index = df["prompt_id"] + "  " + df["category"]
data.columns = [c.upper() for c in model_columns]


from matplotlib.colors import ListedColormap
my_colors = ListedColormap(["#4a8fe0", "#f7b267", "#ee7b30", "#c1272d"])

plt.figure(figsize = (8,12))
sns.heatmap(data, annot=True, fmt="d", cmap=my_colors, vmin=0, vmax=3,
            linewidths=0.5, linecolor = "white")
plt.title("Compliance score by prompt and model")
plt.tight_layout()
plt.savefig("outputs/heatmap_scores.png", dpi=200)
plt.show()
```

```python
pair_id_map = {"P3" : "ddos", "P4" : "ddos",
               "P9" : "brute_force", "P16" : "brute_force",
               "P19":"keylogger", "P20":"keylogger",
               "P21":"ransomware","P22":"ransomware",
               "P25":"disinformation","P26":"disinformation",
               "P15A":"vishing","P15B":"vishing",
               "P28":"osint", "P29":"osint",
               "P32":"rfid","P33":"rfid"}
type_map = {"ddos":"reframe","brute_force":"reframe","keylogger":"reframe","ransomware":"reframe",
            "disinformation":"reframe","vishing":"claim_removed","osint":"subject_swap","rfid":"format_swap"}

df["pair_id"] = df["prompt_id"].map(pair_id_map)
df["pair_type"] = df["pair_id"].map(type_map)
```

```python
rows = []
for pair_name in df["pair_id"].dropna().unique():
    two = df[df["pair_id"] == pair_name]
    first = two.iloc[0]
    second = two.iloc[1]
    row = {"pair":pair_name}
    for m in model_columns:
        row[m] = second[m]-first[m]
    rows.append(row)

score_changes = pd.DataFrame(rows).set_index("pair")
score_changes
```

```python
print(df["pair_type"].isna().sum())
```

```python
sc_colors = ListedColormap(["#2a78d6", "#6fa3e8", "#c4d9f4", "#f2f2f2",
                               "#f8cdb8", "#ee7b30", "#c1272d"])
plt.figure(figsize = (8,5))
sns.heatmap(score_changes, annot = True, fmt = "d" , cmap=sc_colors, center=0, vmin = -3, vmax =3,
            linewidth = 0.5, linecolor="white")
plt.title("Score change within pairs(second - first)")
plt.ylabel("")
plt.savefig("outputs/pair_changes.png", dpi =200)
plt.show()
```

```python
full_compliance = (df[model_columns]==3).sum().sort_values(ascending = False)
print(full_compliance)
full_compliance.plot(kind = "bar", color="#c1272d", figsize = (7,4))
plt.ylabel("prompts that scored 3(full compliance) out of 34")
plt.title("Full compliance per model")
plt.xticks(rotation=0)
plt.ylim(0,34)
plt.tight_layout()
plt.savefig("outputs/full_compliance.png", dpi = 200)
plt.show()
```

```python
counts = pd.DataFrame(index=[0,1,2,3])
for m in model_columns:
    counts[m] = df[m].value_counts()
counts = counts.fillna(0)
counts = counts.sort_index()
counts = counts.astype(int)
counts = counts.T
print(counts)

counts.plot(kind = "bar", stacked = True, colormap= my_colors, figsize = (8,5))
plt.ylabel("number of prompts (N=34)")
plt.xticks(rotation = 0)
plt.legend(title = "score", bbox_to_anchor=(1.02,1), loc="upper left")
plt.tight_layout()
plt.savefig("outputs/score_counts.png", dpi = 200)
plt.show()
```
